```python
# Data Processing
transactions = []
for i in range(0, 7501):
    transactions.append([str(df.values[i, j]) for j in range(0, 20)])

# Training the dataset
from apyori import apriori

rules = apriori(
    transactions,
    min_support=0.003,        # (3 items * 7 days in a week) / 7500 total rows
    min_confidence=0.2,
    min_lift=3,
    min_length=2
)

# Visualizing the results
a = list(rules)
result = [[list(a[i][0])[0]] for i in range(0, len(a))]
```
